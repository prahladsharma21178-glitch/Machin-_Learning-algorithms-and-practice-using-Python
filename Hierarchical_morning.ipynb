```python
import pandas as pd
from sklearn.datasets import load_iris
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import AgglomerativeClustering
from sklearn.metrics import silhouette_score
```

```python
iris=load_iris()
df=pd.DataFrame(iris.data,columns=iris.feature_names)
```

```python
df
scaled=StandardScaler()
X_scaled=scaled.fit_transform(df)
X_scaled
```

```python
l1=[]
for l in ['ward','average','complete','single']:
  hc=AgglomerativeClustering(n_clusters=3,linkage=l)
  a=hc.fit_predict(X_scaled)
  performance=silhouette_score(X_scaled,a)
  l1.append(performance)

final=max(l1)
l1
```

```python
plt.scatter(X_scaled[:,0],
            X_scaled[:,1],
            c=a,
            cmap='rainbow')
plt.xlabel('sepal length')
plt.ylabel('sepal width')
plt.title('hc')
```

```python
performance=silhouette_score(X_scaled,a)
performance
```

score around +1   :  cluster are well separate
score 0.5 to 0.7  :  good clustering
score 0.25 to 0.5 :  moderate clustering
score around 0:     clusters are overlaped
