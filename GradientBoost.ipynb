```python
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.model_selection import train_test_split
from sklearn.datasets import load_breast_cancer
import pandas as pd
from sklearn.metrics import confusion_matrix,accuracy_score,precision_score,recall_score,f1_score
```

```python
d=load_breast_cancer()
df=pd.DataFrame(d.data,columns=d.feature_names)
df['target']=d.target
df.head()
```

```python
X=df.drop('target',axis=1)
Y=df['target']
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.3,random_state=42)
```

```python
GB=GradientBoostingClassifier(n_estimators=100,learning_rate=0.5,max_depth=1,random_state=42)
GB.fit(X_train,Y_train)
```

```python
GB.score(X_train,Y_train)
```

```python
GB.score(X_test,Y_test)
```

```python
p=GB.predict(X_test)
```

```python
confusion_matrix(Y_test,p)
```

** XGboost**

```python

from xgboost import XGBClassifier
from sklearn.model_selection import train_test_split
from sklearn.datasets import load_breast_cancer
import pandas as pd
from sklearn.metrics import confusion_matrix,accuracy_score,precision_score,recall_score,f1_score
```

```python
data=load_breast_cancer()
df=pd.DataFrame(data.data,columns=data.feature_names)
df['target']=data.target
df.head()
X=df.drop('target',axis=1)
Y=df['target']
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.3,random_state=42)
```

```python
XGB=XGBClassifier(n_estimators=100,learning_rate=0.5,max_depth=1,random_state=42,subsample=0.8,colsample=0.8,eval_metric='logloss',reg_lambda=1)
XGB.fit(X_train,Y_train)
```

```python
XGB.score(X_train,Y_train)
```

```python
XGB.score(X_test,Y_test)
```
