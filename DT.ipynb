**DTC**

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
import pandas as pd
from sklearn.metrics import confusion_matrix,accuracy_score,classification_report
from sklearn.preprocessing import LabelEncoder
from sklearn.linear_model import Ridge,Lasso
```

```python
df=pd.read_csv('/content/winequalityN.csv')
df
```

```python
L=LabelEncoder()
df['type']=L.fit_transform(df['type'])
df.info()
df['fixed acidity']=df['fixed acidity'].fillna(df['fixed acidity']).mean()
df['volatile acidity']=df['volatile acidity'].fillna(df['volatile acidity']).mean()
df['citric acid']=df['citric acid'].fillna(df['citric acid']).mean()
df['residual sugar']=df['residual sugar'].fillna(df['residual sugar']).mean()
df['chlorides']=df['chlorides'].fillna(df['chlorides']).mean()
#df['free sulfur dioxide']=df['free sulfur dioxide'].fillna(df['free_sulfur_dioxide']).mean()
df['pH']=df['pH'].fillna(df['pH']).mean()
df['sulphates']=df['sulphates'].fillna(df['sulphates']).mean()
df.info()
```

```python
X=df.drop('type',axis=1)
Y=df['type']
X
Y
```

```python
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.3,random_state=42)
```

```python
DTC=DecisionTreeClassifier(criterion='entropy',max_depth=4)
DTC.fit(X_train,Y_train)
```

**RFC**

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
import pandas as pd
from sklearn.metrics import confusion_matrix,accuracy_score,classification_report
```

```python
d=load_breast_cancer()
df1=pd.DataFrame(d.data,columns=d.feature_names)
df1['cancer']=d.target
```

```python
X=df1.drop('cancer',axis=1)
Y=df1['cancer']
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.3,random_state=42)
```

```python
RFC=RandomForestClassifier(n_estimators=100,max_depth=4)
RFC.fit(X_train,Y_train)
```

```python
RFC.score(X_train,Y_train)
#RFC.score(X_test,Y_test)
```
